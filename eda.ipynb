```python
import httpx
import os
from dotenv import load_dotenv
import pandas as pd
from datetime import datetime, timedelta
```

```python
today = datetime.now().date()
start_of_week = today - timedelta(days=7)  # Monday
end_of_week = start_of_week + timedelta(days=6)
```

```python
load_dotenv()

M2M_TOKEN = os.getenv("M2M_TOKEN")
url = "https://goteamup.com/api/v2/reports/attendances.grouped/data"

headers = {
    "Accept": "application/json",
    "Authorization": f"Bearer {M2M_TOKEN}",
    "TeamUp-Request-Mode": "provider"
}

# We omit the status parameter completely to avoid empty server filtering
params = [
    ("columns", "customer"),
    ('attendance_status[]', 'attended'),
    ("event_start_gte", start_of_week.strftime("%Y-%m-%dT00:00:00Z")),  # Adjust to any date before your business opened
    ("event_start_lte", end_of_week.strftime("%Y-%m-%dT23:59:59Z"))     # Captures all historical logs up to the future cutoff
]

with httpx.Client() as client:
    response = client.get(url, headers=headers, params=params, timeout=30.0)

if response.status_code != 200:
    print(f"Server Error Status: {response.status_code}")
    print(f"Raw Server Message: '{response.text}'")
else:
    report_data = response.json()
```

```python
print(report_data)
```

```python
customer_attendances = report_data.get("rows")
print(customer_attendances)
```

```python
df = pd.DataFrame(customer_attendances)
```
